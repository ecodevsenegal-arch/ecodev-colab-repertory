```
# Définition d'une variable ou affectation
n=int(input("Entrer un nombre : "))
# Début d'une boucle for: répéter un bloc pour chaque élément
for i in range(1,11):
# Affichage d'un résultat à l'écran
    print(n, "x",i,"=", n*i)
```
